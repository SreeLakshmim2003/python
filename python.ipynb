```python
n=int(input("Enter number of steps:"))
for i in range(1,n+1):
    for j in range(1,i+1):
        print(i*j,end=" ")
    print()
```

```python
n=5
for i in range(1,n+1):
    for j in range(i):
        print("*",end="")
    print()

for i in range(n-1,0,-1):
    for j in range(i):
        print("*",end="")
    print()
```

```python
square=lambda a:a*a
rectangle=lambda l,b:l*b
triangle=lambda b,h:0.5*b*h
print("Area of square:",square(5))
print("Area of rectangle:",rectangle(2,6))
print("Area of triangle:",triangle(4,8))
```

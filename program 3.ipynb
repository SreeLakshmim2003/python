```python
names=(input("Enter the 1st name:"))
count=0
for name in names:
    count+=name.lower().count('a')
print("Occurrences of a:",count)
```

```python
color_list1=["red","blue","rose","pink"]
color_list2=["red","yellow"]
for color in color_list1:
    if color not in color_list2:
        print(color)
```

```python
d={'b':10,'a':5,'d':20,'c':30}
print("Ascending order:",dict(sorted(d.items())))
print("Descending order:",dict(sorted(d.items(),reverse=True)))
```

```python
a=list(map(int,input("Enter 1st list").split()))
b=list(map(int,input("Enter 2nd list").split()))
if len(a)==len(b):
    print(" lists have same length")
else:
    print("Lists have different length")
if sum(a)==sum(b):
    print(" lists have same sum")
else:
    print("Lists different sum")
for i in (a):
    if i in(b):
        print("common values",i)
```
